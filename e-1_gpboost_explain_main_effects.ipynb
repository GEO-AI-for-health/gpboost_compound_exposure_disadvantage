```python
import os
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt
import seaborn as sns
from pygam import LinearGAM, s
import warnings
from matplotlib.lines import Line2D

warnings.filterwarnings('ignore')

# Set plot style
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 20
plt.rcParams['axes.titlesize'] = 20
plt.rcParams['axes.labelsize'] = 20
plt.rcParams['xtick.labelsize'] = 17
plt.rcParams['ytick.labelsize'] = 17
plt.rcParams['legend.fontsize'] = 24


out_dir = 'shap_outputs'
os.makedirs(out_dir, exist_ok=True)

# Define feature names: raw (for data selection) and display (for plotting)
feature_names_raw = [
    'household_income', 'age', 'gender', 'education_level', 'employment_status', 'neighborhood_type',
    'work_study', 'housework', 'personal_affair', 'leisure', 'travel', 'transportation', 'residence',
    'industry', 'company', 'shopping', 'restaurant', 'life_service', 'education_culture', 'entertainment',
    'sport_fitness', 'recreation_tourism', 'healthcare', 'workday', 'time_hour', 'mobility_status',
    'time_nonflexibility', 'home_ornot', 'POI_density', 'POI_diversity', 'tree_height'
]
feature_names_display = [
    'Household income', 'Age', 'Gender', 'Education level', 'Employment status', 'Neighborhood type',
    'Work_study', 'Housework', 'Personal affair', 'Leisure', 'Travel', 'Transportation',
    'Residence', 'Industry', 'Company', 'Shopping', 'Restaurant', 'Life service',
    'Education culture', 'Entertainment', 'Sport_fitness', 'Recreation_tourism', 'Healthcare',
    'Workday', 'Time hour', 'Mobility status', 'Time inflexibility', 'Being neighborhood',
    'POI density', 'POI diversity', 'Tree height'
]


bootstrap_main_effects = np.load(os.path.join(out_dir, 'bootstrap_main_effects.npy'))
main_effects_mean = np.load(os.path.join(out_dir, 'main_effects_mean.npy'))
main_effects_low = np.load(os.path.join(out_dir, 'main_effects_low.npy'))
main_effects_up = np.load(os.path.join(out_dir, 'main_effects_up.npy'))
print(f"Loaded SHAP arrays from {out_dir}")


data = pd.read_csv('C:/Files/GLAN_DATA/first_wave/downscale_data_60/data_with_activity_sequence_all_recod_downsampling_activity_travel.csv')
data = data[data['time_weight'] >= 5]
data = data[data['tree_height'] >= 0]
fixed_sample_indices = np.load(os.path.join('models', 'selected_index.npy'))
X_data = data[feature_names_raw].values  
X_data_sample = X_data[fixed_sample_indices]
X_data_sample_df = pd.DataFrame(X_data_sample, columns=feature_names_display)
print(f"X_data_sample_df shape: {X_data_sample_df.shape}")

def smoother(X, y):
    lam = np.arange(100, 301, 20).reshape(-1, 1)
    try:
        gam = LinearGAM(s(0), fit_intercept=False).gridsearch(X, y, lam=lam)
        XX = gam.generate_X_grid(term=0)
        pdep, _ = gam.partial_dependence(term=0, X=XX, width=0.95)
        return XX, pdep
    except Exception as e:
        print(f"Smoothing error: {str(e)}")
        return None, None

def plot_feature_dependence(ax, feat_name):
    feat_idx = feature_names_display.index(feat_name)
    df = pd.DataFrame({
        'd': X_data_sample_df[feat_name].values,
        'shap_val': main_effects_mean[:, feat_idx]
    })
   
    df['d_jitter'] = df['d'] + np.random.uniform(-0.05, 0.05, size=df['d'].shape)
    
    nonsig = np.logical_and(main_effects_low[:, feat_idx] <= 0, main_effects_up[:, feat_idx] >= 0)
    
    sns.scatterplot(ax=ax, x='d_jitter', y='shap_val', data=df[nonsig], color='grey', alpha=0.8, s=50)
    sns.scatterplot(ax=ax, x='d_jitter', y='shap_val', data=df[~nonsig], color=shap.plots.colors.blue_rgb, alpha=0.8, s=50)

    order = np.argsort(df['d'].values)
    sorted_x = df['d'].values[order].reshape(-1, 1)
    sorted_mean = df['shap_val'].values[order].reshape(-1, 1)
    sorted_low = main_effects_low[:, feat_idx][order].reshape(-1, 1)
    sorted_up = main_effects_up[:, feat_idx][order].reshape(-1, 1)

    a, b = smoother(sorted_x, sorted_mean)
    if a is not None:
        ax.plot(a, b, '--', color='red', linewidth=2, label='GAM trend')
        a1, b1 = smoother(sorted_x, sorted_low)
        a2, b2 = smoother(sorted_x, sorted_up)
        if a1 is not None and a2 is not None:
            ax.fill_between(a1.reshape(-1), b1, b2, alpha=0.2, color='grey', linewidth=0, label='95% CI')
        else:
            ax.text(0.5, 0.5, 'Unable to compute CI', ha='center', va='center', transform=ax.transAxes, color='red')
    else:
        ax.text(0.5, 0.5, 'Insufficient data for GAM smoothing', ha='center', va='center', transform=ax.transAxes, color='red')

    ax.axhline(y=0, color='black', linestyle='--', linewidth=1)
    ax.set_ylabel('Prediction Contribution')
    ax.set_xlabel(feat_name)

def build_legend_handles():
    handle_points_nonsig = Line2D([0], [0], marker='o', color='none',
                                  markerfacecolor='grey', alpha=0.8, markersize=12, label='Non-significant')
    handle_points_sig = Line2D([0], [0], marker='o', color='none',
                               markerfacecolor=shap.plots.colors.blue_rgb, alpha=0.8, markersize=12, label='Significant')
    handle_trend = Line2D([0], [0], linestyle='--', color='red', linewidth=4, label='GAM trend')
    handle_ci = Line2D([0], [0], linestyle='-', color='grey', alpha=0.5, linewidth=12, label='95% CI')
    return [handle_points_nonsig, handle_points_sig, handle_trend, handle_ci]
```

```python

from matplotlib.ticker import FixedLocator, FixedFormatter

features = ['Household income', 'Age', 'Gender', 'Education level', 'Employment status', 'Neighborhood type']
title = "Socioeconomic status"
filename = "main_effects_dependence_Socioeconomic_status.pdf"
cols = 3
num = len(features)
rows = (num + cols - 1) // cols


fig_w = 7.8 * cols   
fig_h = 6.0 * rows   
fig, axes = plt.subplots(rows, cols, figsize=(fig_w, fig_h), dpi=1000)
axes = axes.flatten() if rows > 1 or cols > 1 else [axes]


custom_ticks = {
    'Household income': {
        'locs': [1, 2, 3],
        'labels': ['Less 20k', '20k - 40k', 'Over 40k'],
        'xlabel': 'Household income (HKD/month)'
    },
    'Age': {
        'locs': [1, 2, 3],
        'labels': ['18 - 30', '31 - 45', '46 - 66'],
        'xlabel': 'Age (years)'
    },
    'Gender': {
        'locs': [0, 1],
        'labels': ['Female', 'Male'],
        'xlabel': 'Gender'
    },
    'Education level': {
        'locs': [1, 2, 3],
        'labels': ["Below Bachelor's\ndegree",
                   "Bachelor's\ndegree",
                   "Master's or\nDoctor's degree"],
        'xlabel': 'Education level'
    },
    'Employment status': {
        'locs': [1, 2, 3],
        'labels': ['The unemployed/\nretirees/homemakers',
                   'Part-time employment/\nthe self-employed',
                   'Full-time\nemployment'],
        'xlabel': 'Employment status'
    },
    
    'Neighborhood type': {
        'locs': [1, 2, 3],
        'labels': ['Private\nhousing neighborhood',
                   'Mixed neighborhood',
                   'Public\nhousing neighborhood'],
        'xlabel': 'Neighborhood type'
    },
    
}

for i, feat in enumerate(features):
    ax = axes[i]
    plot_feature_dependence(ax, feat)  

    
    ax.set_ylabel('Prediction Contribution')

    
    if feat in custom_ticks:
        info = custom_ticks[feat]
        ax.xaxis.set_major_locator(FixedLocator(info['locs']))
        ax.xaxis.set_major_formatter(FixedFormatter(info['labels']))
        ax.set_xlabel(info['xlabel'])
        
        ax.tick_params(axis='x', rotation=0, pad=6)
    else:
        ax.set_xlabel(feat)


for j in range(num, len(axes)):
    fig.delaxes(axes[j])


handles = build_legend_handles()
fig.legend(handles=handles, loc='center left', bbox_to_anchor=(1.01, 0.5), frameon=False)

fig.suptitle(title)
plt.tight_layout(rect=[0, 0.02, 0.85, 0.96])

save_path = os.path.join(out_dir, filename)
plt.savefig(save_path, format='pdf', bbox_inches='tight')
plt.show()
print(f"Saved: {save_path}")
```

```python

from matplotlib.ticker import FixedLocator, FixedFormatter

features = ['Work_study', 'Housework', 'Personal affair', 'Leisure', 'Travel']
title = "Activity type contexts"
filename = "main_effects_dependence_Activity_type_contexts.pdf"


cols = 3
num = len(features)
rows = (num + cols - 1) // cols

fig_w = 6.0 * cols  
fig_h = 4.8 * rows
fig, axes = plt.subplots(rows, cols, figsize=(fig_w, fig_h), dpi=800)
axes = axes.flatten() if rows > 1 or cols > 1 else [axes]


binary_tick_labels = {
    'Work_study': ('No', 'Yes'),
    'Housework': ('No', 'Yes'),
    'Personal affair': ('No', 'Yes'),
    'Leisure': ('No', 'Yes'),
    'Travel': ('No', 'Yes'),
}

for i, feat in enumerate(features):
    ax = axes[i]
    plot_feature_dependence(ax, feat)  

    
    ax.set_ylabel('Prediction Contribution', labelpad=8)

    
    labs = binary_tick_labels.get(feat, ('0', '1'))
    ax.xaxis.set_major_locator(FixedLocator([0, 1]))
    ax.xaxis.set_major_formatter(FixedFormatter(list(labs)))
    ax.set_xlabel(feat, labelpad=8)  
    ax.tick_params(axis='x', rotation=0, pad=6)


for j in range(num, len(axes)):
    fig.delaxes(axes[j])


handles = build_legend_handles()
fig.legend(handles=handles, loc='center left', bbox_to_anchor=(1.01, 0.5), frameon=False)

fig.suptitle(title)
plt.tight_layout(rect=[0, 0.02, 0.82, 0.96])  

save_path = os.path.join(out_dir, filename)
plt.savefig(save_path, format='pdf', bbox_inches='tight')
plt.show()
print(f"Saved: {save_path}")
```
